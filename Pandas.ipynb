 Assignment Questions of Pandas :

```python
import pandas as pd
from pandas import Series
```

```python
## Section A: Easy-

sales = Series([120, 150, 180, 210])

marks = Series([72, 81, 69, 90])
print(marks.values)

print(marks.index)

temperatures = Series([28, 31, 30, 29], index=['Mon', 'Tue', 'Wed', 'Thu'])
print(temperatures)

population = Series([35, 42, 31], index=['Delhi', 'Mumbai', 'Jaipur'])
print(population['Jaipur'])

score = Series([35, 68, 44, 79, 52])
print(score[score > 50])

subjects = Series([76, 83, 91], index=['Physics', 'Chemistry', 'Maths'])
print('Chemistry' in subjects.index)

print(subjects.to_dict())

prices = {'Pen': 10, 'Notebook': 55, 'Bag': 700}
price_series = Series(prices)
print(price_series)

print("The dictionary keys become the Series index")

data = {'A': 10, 'B': 20, 'C': 30}
s = Series(data, index=['A', 'B', 'C', 'D'])
print(s[s.isnull()].index[0])

print(pd.isnull(s))

print(pd.notnull(s))

sales.name = 'Monthly Sales'

sales.index.name = 'Month'

## Section B: Medium

s = Series([18, 45, 27, 61], index=['P', 'Q', 'R', 'S'])
print(s[s > 30])

s = Series([500, 600, 700], index=['East', 'West', 'North'])
print('West' in s)
print(500 in s)

city_data = {'Delhi': 80, 'Pune': 65, 'Kochi': 70}
city = Series(city_data, index=['Delhi', 'Kochi', 'Pune', 'Chennai'])
print(city)
print(city[city.isnull()].index[0])

s = Series({'W': 5, 'X': 10, 'Y': 15}, index=['W', 'X', 'Y', 'Z'])
print(pd.isnull(s))

print(pd.notnull(s))

s1 = Series([10, 20, 30], index=['A', 'B', 'C'])
s2 = Series([1, 2, 3], index=['C', 'A', 'B'])
print(s1 + s2)

s1 = Series({'A': 12, 'B': 18})
s2 = Series({'B': 5, 'C': 9})
print(s1 + s2)
print("Addition aligns on index labels; A and C exist in only one Series, so they become NaN")

stock = Series([25, 40, 18], index=['Item1', 'Item2', 'Item3'])
stock_dict = stock.to_dict()
stock2 = Series(stock_dict)
print(stock2)

marks = {'Maths': 88, 'Science': 74, 'English': 69, 'History': 81}
marks_series = Series(marks, index=['History', 'English', 'Science', 'Maths'])
print(marks_series[marks_series > 70])

rev = Series([300, 450, 380], index=['North', 'South', 'East'])
rev.name = 'Revenue'
rev.index.name = 'Branch'
print(rev)

## Section C : Hard:

north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
print(total)

d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']
result = Series(d, index=labels)
print(result[result.isnull()].index[0])
print(pd.isnull(result)['Yellow'])
print("Yellow is in the index but not in the dictionary, so no value exists for it")
print(result[result.notnull()].index.tolist())

s = Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = Series(d)
print(s2.index.tolist())
print(s2.values.tolist())
print(s.equals(s2))

products = ['Laptop', 'Phone', 'Tablet', 'Watch']
product_sales = Series([850, 600, 450, 300], index=products)
print(product_sales.values)
print(product_sales.index)
print(product_sales['Phone'])
print(product_sales[product_sales > 500])
print('Tablet' in product_sales.index)
print(product_sales.to_dict())

data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print(s.index.tolist())
print(s[s.isnull()].index[0])
print(s[s > 20])
print(pd.notnull(s))
print(s.name, s.index.name)
```
