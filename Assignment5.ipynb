Data Frames Assignment :

Section A : Easy

```python
import pandas as pd
import numpy as np

student_df = pd.DataFrame({'Name': ['Amit', 'Riya', 'Karan', 'Neha'], 'Age': [18, 19, 18, 20], 'City': ['Delhi', 'Mumbai', 'Pune', 'Jaipur']})

print(student_df)

print(student_df.columns)

print(student_df.Name)

print(student_df['City'])

new_df = student_df[['Name', 'City']]
print(new_df)

book_df = pd.DataFrame({'Book': ['Wings of Fire', 'Godan', 'Gitanjali'], 'Author': ['APJ Abdul Kalam', 'Premchand', 'Tagore'], 'Price': [250, 180, 150]})
print(book_df)

book_df['Publisher'] = 'Penguin'
print(book_df)

df9 = pd.DataFrame({'Item': ['A', 'B', 'C', 'D', 'E'], 'Qty': [10, 20, 30, 40, 50]})
df9['SerialNo'] = np.arange(5)
print(df9)

del df9['SerialNo']
print(df9)

movie_df = pd.DataFrame({'Movie': ['Dangal', '3 Idiots', 'Lagaan'], 'Year': [2016, 2009, 2001], 'Rating': [8.4, 8.4, 8.1]})
print(movie_df)

movie_new = pd.DataFrame(movie_df, columns=['Movie', 'Rating', 'Director'])
print(movie_new)

city_series = pd.Series(['Delhi', 'Mumbai'], index=[0, 3])
print(city_series)

df14 = pd.DataFrame({'Manager': ['M1', 'M2', 'M3', 'M4']})
df14['Branch'] = city_series
print(df14)


```

Section B: Medium

```python

employee_df = pd.DataFrame({'Employee': ['Ravi', 'Sita', 'Mohan', 'Anita', 'Vijay'], 'Department': ['HR', 'IT', 'Sales', 'IT', 'Finance'], 'Experience': [5, 3, 8, 2, 10]})
emp_small = employee_df[['Employee', 'Experience']]
print(emp_small)

print(employee_df['Department'])
print(employee_df.columns)

employee_df['Office'] = 'Head Office'
print(employee_df)

employee_df['Office'] = np.arange(5)
print(employee_df)

office_names = pd.Series(['Delhi', 'Mumbai', 'Pune'], index=[0, 2, 4])
employee_df['Office'] = office_names
print(employee_df)

del employee_df['Office']
print(employee_df.columns)

product_df = pd.DataFrame({'Product': ['Pen', 'Notebook', 'Bag'], 'Category': ['Stationery', 'Stationery', 'Accessories'], 'Price': [10, 50, 500]})
reordered_df = pd.DataFrame(product_df, columns=['Price', 'Product', 'Category'])
print(reordered_df)

source_df = pd.DataFrame({'Name': ['A', 'B', 'C'], 'Age': [20, 21, 22], 'City': ['X', 'Y', 'Z']})
four_col_df = pd.DataFrame(source_df, columns=['Name', 'Age', 'City', 'Country'])
print(four_col_df)
print(four_col_df['Country'])

marks_df = pd.DataFrame({'Name': ['Aman', 'Bhavna', 'Chetan'], 'Total Marks': [450, 480, 420]})
print(marks_df['Total Marks'])

clip_df = pd.read_clipboard()
print(clip_df.columns)
three_col_df = clip_df[[clip_df.columns[0], clip_df.columns[1], clip_df.columns[2]]]
print(three_col_df)
```

Section C : Hard:

```python

course_df = pd.DataFrame({'Student': ['Arjun', 'Bina', 'Chirag', 'Divya', 'Esha'], 'Course': ['Python', 'Java', 'Python', 'C++', 'Java'], 'Score': [85, 78, 92, 66, 88]})
print(course_df.columns)
print(course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
print(student_score_df)
course_df['Batch'] = 'Batch-1'
print(course_df)

travel_df = pd.DataFrame({'Place': ['Goa', 'Shimla', 'Manali', 'Agra', 'Jaipur'], 'Days': [3, 4, 5, 2, 3]})
travel_df['Code'] = np.arange(5)
print(travel_df)
travel_df['Code'] = pd.Series([101, 104], index=[1, 4])
print(travel_df)

school_df = pd.DataFrame({'Student': ['Rohan', 'Sneha', 'Tanvi', 'Uday'], 'Class': [10, 10, 9, 9], 'Section': ['A', 'B', 'A', 'C']})
school_house_df = pd.DataFrame(school_df, columns=['Student', 'Class', 'Section', 'House'])
print(school_house_df)
school_house_df['House'] = 'Blue'
print(school_house_df)

inventory_df = pd.DataFrame({'Item': ['Chair', 'Table', 'Lamp', 'Shelf'], 'Stock': [20, 10, 35, 8], 'Price': [1500, 4000, 600, 2500]})
inventory_df['Supplier'] = pd.Series(['S1', 'S2'], index=[0, 2])
print(inventory_df)
del inventory_df['Supplier']
print(inventory_df)


```
