```python
import numpy as np

# Marks of 10 students in 3 subjects
marks = np.array([
    [78, 85, 90],
    [45, 50, 55],
    [30, 40, 35],
    [88, 92, 80],
    [60, 65, 70],
    [25, 30, 28],
    [95, 90, 98],
    [55, 58, 52],
    [40, 45, 48],
    [70, 75, 80]
])

# Total marks of each student
total = np.sum(marks, axis=1)

# Average marks of each student
average = np.mean(marks, axis=1)

# Students with average >= 50
qualified = np.where(average >= 50)

print("Total Marks:")
print(total)

print("\nAverage Marks:")
print(average)

print("\nStudents with Average >= 50:")
print(qualified[0])  # Student indices
```

```python
import numpy as np

# Create a 4×4 matrix with numbers 1 to 16
matrix = np.arange(1, 17).reshape(4, 4)

print("Original Matrix:")
print(matrix)

# Transpose
transpose_matrix = matrix.T
print("\nTranspose:")
print(transpose_matrix)

# Row-wise sums
row_sums = np.sum(matrix, axis=1)
print("\nRow-wise Sums:")
print(row_sums)

# Column-wise sums
col_sums = np.sum(matrix, axis=0)
print("\nColumn-wise Sums:")
print(col_sums)

# Mean
mean_value = np.mean(matrix)
print("\nMean:", mean_value)

# Maximum and Minimum
max_value = np.max(matrix)
min_value = np.min(matrix)

print("Maximum:", max_value)
print("Minimum:", min_value)

# Central 2×2 matrix
central_matrix = matrix[1:3, 1:3]
print("\nCentral 2×2 Matrix:")
print(central_matrix)
```
