# Section B: Moderate

NumPY:

```python

sales = np.array([
    [120, 150, 180],
    [140, 160, 175],
    [135, 155, 190],
    [150, 170, 200]
])
print(sales.sum(axis=1))
print(sales.sum(axis=0))
print(sales.sum(axis=1).argmax() + 1)

a22 = np.arange(10, 46).reshape(6, 6)
sub = a22[2:6, 1:5]
print(sub)
print(sub.mean())

weights = np.array([48, 62, 75, 81, 54, 69, 90])
print(weights[weights > 70])
new_weights = np.where(weights < 55, 55, weights)
print(new_weights)
print(weights)

quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
cost = quantity * price
bill = cost.sum()
print(cost)
print(bill)
print(cost / bill * 100)

a25 = np.arange(1, 13).reshape(3, 4)
print(a25.var(axis=1), a25.std(axis=1))
print(a25.var(axis=0), a25.std(axis=0))
print("axis=0 operates down the rows, giving one result per column")
print("axis=1 operates across the columns, giving one result per row")

m1 = np.array([[1, 2, 3], [4, 5, 6]])
m2 = np.array([[7, 8], [9, 10], [11, 12]])
print(m1 @ m2)
print(m1.shape, m2.shape)
print("Element-wise multiplication needs equal or broadcast-compatible shapes; (2,3) and (3,2) are neither")

rng = np.random.default_rng(42)
r = rng.integers(10, 100, 25)
sorted_r = np.sort(r)
print(sorted_r)
print(sorted_r[-5:][::-1])
print(np.unique(r).size)

a = np.array([5, 10, 15, 20, 25, 30])
b = a[1:5]
c = a[1:5].copy()
b[1] = 99
c[2] = 77
print(a)
print(b)
print(c)
print("A view shares memory with the original so changes show in both; a copy is independent")
```

# Section C : Hard

NumPY:

```python

rng = np.random.default_rng(42)
marks = rng.integers(20, 101, size=(12, 4))
print(marks)
avg = marks.mean(axis=1)
print(avg)
print(marks.mean(axis=0))
print(marks.std(axis=0))
print((marks >= 40).all(axis=1).sum())
print(np.where(avg >= 75)[0] + 1)
print(marks.std(axis=0).argmax() + 1)
print("Students are numbered 1 to 12, tests 1 to 4")

m = np.arange(21, 46).reshape(5, 5)
center = m[1:4, 1:4]
ct = center.T
prod = center @ ct
print(m.shape, center.shape, ct.shape, prod.shape)
print(center)
print(ct)
print(prod)
print(prod.sum(axis=1))
mx = prod.max()
print(mx)
print(np.argwhere(prod == mx))

rng = np.random.default_rng(7)
qty = rng.integers(10, 41, size=(6, 4))
prices = np.array([150, 240, 90, 320])
revenue = qty * prices
print(revenue.sum(axis=1))
print(revenue.sum(axis=0))
print(revenue.sum(axis=1).argmax() + 1)
print(qty.sum(axis=0).argmax() + 1)
new_prices = np.where(qty.sum(axis=0) > 100, prices * 0.9, prices)
print(new_prices)
print(prices)
print((qty * new_prices).sum(axis=0))
print((qty * new_prices).sum())

rng = np.random.default_rng(1)
data = rng.normal(50, 12, 100)
print(((data < 30) | (data > 70)).sum())
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))
for name, arr in [('original', data), ('cleaned', cleaned)]:
    print(name, arr.mean(), arr.var(), arr.std(), arr.min(), arr.max())
print("Clipping pulls extremes inward, so variance, standard deviation and range fall, while the mean shifts only slightly")
```

Pandas:

```python

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

total = branch_a + branch_b + branch_c
print(total)
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print(common.tolist())
print(total[common])
filled = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print(filled)
print("Ordinary addition marks any quarter missing in a branch as NaN (unknown); fill_value=0 treats absence as no earnings")

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print(s)
print(s[s.isnull()].index.tolist())
valid = s[s >= 70]
print(valid)
d = valid.to_dict()
print(d)
rebuilt = pd.Series(d)
print(rebuilt)
print((rebuilt == valid[rebuilt.index]).all())
print("A missing record means no data, not a score of zero; zero would wrongly lower averages and imply the student scored nothing")
```
